```python
import os
import numpy as np
from matplotlib import pyplot as plt
import seaborn as sns
import pandas as pd
from datetime import datetime
sns.set_theme(style="whitegrid")
```

```python
models = [
    r"D:\Projects\whakaari\output_bak\models\fm_2.00_0.75_2.00_dsar-hf-mf-rsam\all.fts",
    r"D:\Projects\whakaari\output_bak_new\models\fm_2.00_0.75_2.00_dsar-hf-mf-rsam\all.fts",
    r"D:\Projects\whakaari\all-ws.fts",
    r"D:\Projects\whakaari\all-ws-nn.fts",
]
```

```python
def plot(df, filename: str = None):
    filename = filename if filename else f"{datetime.now()}"
    filename = filename.replace(":", "_")
    sns.set_color_codes("pastel")

    fig, ax = plt.subplots()
    sns.barplot(x="value", y="feature", data=df.head(20),
                label="Most used feature", color="b")

    ax.legend(ncol=2, loc="lower right", frameon=True)
    ax.set(xlabel="Value", ylabel="")
    sns.despine(left=True, bottom=True)

    plt.savefig(f"{filename}.png")
```

```python
def read(file):
    df = pd.read_csv(file, names=['value','feature'])
    return df
```

```python
def main():
    for file in models:
        df = read(file)
        plot(df)
```

```python
main()
```
